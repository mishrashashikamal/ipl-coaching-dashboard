```python
# ============================================
# AI COACHING REPORT GENERATOR
# ============================================
import google.generativeai as genai
from dotenv import load_dotenv
import os

# Load .env with full path
load_dotenv(dotenv_path=r"C:\Users\Shashi Kamal Mishra\OneDrive\Documents\ipl-coaching-dashboard\.env")

# Get key and configure
API_KEY = os.getenv("GEMINI_API_KEY")
genai.configure(api_key=API_KEY)  # ← this line was missing!
model = genai.GenerativeModel("gemini-2.5-flash")

player = "MS Dhoni"

prompt = f"""
You are an expert IPL cricket coach. Based on the following 
data analysis of {player}, generate a professional coaching report.

PLAYER STATS:
- Total Balls Faced: 3947
- Total Runs: 5243
- Strike Rate: 132.84
- Dot Ball %: 36.05%

PHASE-WISE STRIKE RATE:
- Powerplay (0-5): 75.64
- Middle (6-14): 99.01
- Death (15-19): 168.13

DISMISSAL PATTERNS:
- Caught: 92 (62%)
- Run Out: 25 (17%)
- Bowled: 20 (13%)
- Pace dismissals: 79.87%
- Spin dismissals: 20.13%

TOP BOWLERS WHO DISMISSED HIM:
- PP Ojha (Left arm spin): 7 times
- Z Khan (Left arm pace): 7 times
- JJ Bumrah (Right arm pace): 4 times

BEST VENUES:
- Mohali: SR 179.49
- Chinnaswamy: SR 177.01

WORST VENUES:
- Arun Jaitley Stadium Delhi: SR 83.05
- Motera Ahmedabad: SR 88.14

DEATH OVERS PRESSURE:
- Death overs SR: 196.97
- Avg runs per match: 20.02
- Total sixes in death overs: 136

Please provide:
1. Overall player assessment (3-4 lines)
2. Key strengths (3 points)
3. Key weaknesses (3 points)
4. Specific practice drills (4-5 drills)
5. Match strategy recommendations (3 points)
"""

print("Generating AI coaching report... 🤖")
response = model.generate_content(prompt)
print(response.text)
```

```python
# ============================================
# SAVE THE COACHING REPORT
# ============================================

# Save report to a file
with open('dhoni_coaching_report.txt', 'w', encoding='utf-8') as f:
    f.write(f"IPL COACHING REPORT — {player}\n")
    f.write("="*50 + "\n\n")
    f.write(response.text)

print("✅ Report saved to dhoni_coaching_report.txt!")
```
